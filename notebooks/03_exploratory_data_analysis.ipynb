# Phase 2: Exploratory Data Analysis (EDA) & Data Cleaning

In this notebook, we analyze the cleaned housing dataset produced by `src/cleaning.py`:
- **Data Quality Audits:** Verifying null removal, unit conversions, and domain filters.
- **Distribution Analyses:** Examining price and square footage skewness.
- **Bivariate & Multivariate Relationships:** Inspecting how price scales with BHK and area across localities.
- **Feature Correlations:** Identifying multicollinearity between features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('../data/processed/pune_house_data_cleaned.csv')
print(f"Cleaned dataset shape: {df.shape}")
df.head()
```

## 1. Price Distribution & Right-Skewness
Real estate prices are bounded on the left (cannot be negative) and possess an open-ended right tail of high-value properties.

```python
plt.figure(figsize=(9, 4.5))
sns.histplot(df['price'], bins=60, kde=True, color='#2563eb')
plt.title('Distribution of House Prices (Lakhs INR)')
plt.xlabel('Price (₹ Lakhs)')
plt.axvline(df['price'].median(), color='red', linestyle='--', label=f'Median: ₹{df["price"].median():.1f} L')
plt.legend()
plt.show()
```

## 2. Area (total_sqft) vs Price by BHK
Checking linearity and variance spread across bedroom counts.

```python
plt.figure(figsize=(9, 4.5))
sns.scatterplot(data=df[df['bhk'].between(1, 4)], x='total_sqft', y='price', hue='bhk', palette='viridis', alpha=0.6)
plt.title('Total Sq. Ft. vs Price (Colored by BHK)')
plt.xlabel('Area (Sq. Ft.)')
plt.ylabel('Price (₹ Lakhs)')
plt.show()
```

## 3. Correlation Heatmap
Checking collinearity among numerical features.

```python
plt.figure(figsize=(7, 5))
num_cols = ['total_sqft', 'bhk', 'bath', 'balcony', 'price', 'price_per_sqft']
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Matrix')
plt.show()
```
